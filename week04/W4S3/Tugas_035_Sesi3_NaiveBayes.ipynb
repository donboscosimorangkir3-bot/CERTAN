### Soal
Prediksi spesies bunga Iris untuk data baru berikut:

[7.3, 3.8, 4.6, 2.3]

### 1. Import Library

Tahap ini dilakukan untuk mengimpor library yang
dibutuhkan dalam proses klasifikasi. NumPy digunakan
untuk mengolah data dalam bentuk array, Scikit-learn
menyediakan dataset Iris dan algoritma Gaussian
Naïve Bayes yang digunakan dalam praktikum.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB
```

### 2. Memuat Dataset dan Melatih Model

Dataset Iris dimuat menggunakan Scikit-learn.
Selanjutnya, fitur dan label dipisahkan untuk
digunakan dalam proses klasifikasi.

Dataset dibagi menjadi data training dan testing.
Model Gaussian Naïve Bayes kemudian dilatih
menggunakan data training agar dapat mempelajari
pola karakteristik dari setiap spesies bunga.

```python
# Memuat dataset Iris
iris = load_iris()

X = iris.data
y = iris.target

# Membagi data training dan testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=0,
    stratify=y
)

# Melatih model Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)

print("Model Naive Bayes berhasil dilatih.")
```

### 3. Prediksi Data Baru

Pada tahap ini, model Gaussian Naïve Bayes yang
telah dilatih digunakan untuk memprediksi spesies
bunga Iris berdasarkan data baru berikut:

[7.3, 3.8, 4.6, 2.3]

Data tersebut memiliki empat fitur, yaitu panjang
sepal, lebar sepal, panjang petal, dan lebar petal.
Model menghasilkan prediksi berupa label numerik
yang kemudian diubah menjadi nama spesies bunga.

```python
# Data bunga baru
input_data = np.array([[7.3, 3.8, 4.6, 2.3]])

# Melakukan prediksi
prediksi = nb.predict(input_data)

# Mengubah label menjadi nama spesies
nama_spesies = iris.target_names[prediksi[0]]

print("Input data:", input_data[0])
print("Prediksi spesies:", nama_spesies)
```

### 4. Hasil Prediksi dan Pembahasan

Berdasarkan hasil eksekusi program, model Gaussian
Naïve Bayes menghasilkan prediksi spesies untuk

Hasil prediksi menunjukkan spesies Input data: [7.3 3.8 4.6 2.3]
Prediksi spesies: virginica.
