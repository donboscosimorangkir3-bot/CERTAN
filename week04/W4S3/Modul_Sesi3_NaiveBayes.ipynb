### 1. Import Library

Tahap pertama adalah mengimpor library yang diperlukan untuk proses klasifikasi menggunakan algoritma Naïve Bayes. NumPy digunakan untuk mengolah data berbentuk array, Pandas untuk mengelola data, dan Matplotlib untuk membuat visualisasi. Selain itu, library Scikit-learn digunakan untuk memuat dataset Iris, membagi data training dan testing, menerapkan algoritma Gaussian Naïve Bayes, serta mengevaluasi hasil klasifikasi.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

### 2. Memuat Dataset Iris

Pada tahap ini, dataset Iris dimuat menggunakan fungsi `load_iris()` dari Scikit-learn. Dataset Iris berisi data pengukuran bunga yang terdiri dari empat fitur, yaitu panjang sepal, lebar sepal, panjang petal, dan lebar petal. Dataset ini digunakan sebagai data untuk melatih model Naïve Bayes agar dapat memprediksi spesies bunga berdasarkan karakteristik pengukurannya.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

### 3. Mengubah Data Menjadi DataFrame dan Memisahkan Fitur serta Target

Pada tahap ini, data Iris diubah menjadi DataFrame menggunakan Pandas agar lebih mudah dibaca dan dikelola. Empat fitur pengukuran bunga digunakan sebagai kolom, sedangkan nama spesies ditambahkan pada kolom `Species`.

Selanjutnya, data dipisahkan menjadi dua variabel, yaitu `X` sebagai fitur atau karakteristik pengukuran bunga dan `y` sebagai target atau label spesies. Variabel `X` berisi empat fitur pengukuran, sedangkan `y` berisi nama spesies bunga Iris. Data fitur kemudian ditampilkan untuk melihat struktur dan isi dataset.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

### 4. Menampilkan Target atau Label Spesies

Pada tahap ini, variabel `y` ditampilkan untuk melihat label atau target dari dataset Iris. Label tersebut menunjukkan spesies dari setiap data bunga, yaitu `setosa`, `versicolor`, atau `virginica`.

Informasi target ini akan digunakan oleh algoritma Gaussian Naïve Bayes sebagai acuan dalam mempelajari hubungan antara fitur pengukuran bunga dan spesiesnya selama proses pelatihan model.

```python
y
```

### 5. Mengubah Label Spesies Menjadi Data Numerik

Pada tahap ini, `LabelEncoder` dari Scikit-learn digunakan untuk mengubah label spesies bunga yang berbentuk teks menjadi nilai numerik. Algoritma machine learning membutuhkan representasi label yang sesuai agar proses pelatihan dan klasifikasi dapat dilakukan.

Fungsi `fit_transform()` digunakan untuk mempelajari label yang tersedia sekaligus mengubahnya menjadi angka. Label spesies `setosa`, `versicolor`, dan `virginica` akan direpresentasikan dengan nilai numerik yang berbeda. Hasil encoding disimpan kembali ke dalam variabel `y` dan ditampilkan untuk memeriksa hasil konversi.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

### 6. Membagi Data Training dan Testing

Pada tahap ini, dataset dibagi menjadi data training dan data testing menggunakan fungsi `train_test_split()` dari Scikit-learn. Pembagian data dilakukan dengan proporsi 80% untuk data training dan 20% untuk data testing.

Data training digunakan untuk melatih model Gaussian Naïve Bayes agar dapat mempelajari pola hubungan antara fitur pengukuran bunga dan spesiesnya. Sementara itu, data testing digunakan untuk menguji kemampuan model dalam memprediksi data yang belum digunakan selama proses pelatihan.

Parameter `random_state=0` digunakan agar pembagian data menghasilkan hasil yang konsisten ketika kode dijalankan kembali.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

### 7. Melatih Model Gaussian Naïve Bayes

Pada tahap ini, algoritma Gaussian Naïve Bayes diimpor dari Scikit-learn dan digunakan untuk membangun model klasifikasi. Gaussian Naïve Bayes merupakan algoritma klasifikasi yang menggunakan Teorema Bayes dengan asumsi bahwa nilai setiap fitur mengikuti distribusi Gaussian atau distribusi normal.

Objek model dibuat menggunakan `GaussianNB()`. Selanjutnya, fungsi `fit()` digunakan untuk melatih model menggunakan data training, yaitu `X_train` sebagai fitur dan `y_train` sebagai label spesies. Melalui proses pelatihan ini, model mempelajari karakteristik setiap spesies bunga Iris sehingga dapat digunakan untuk memprediksi spesies pada data baru.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

### 8. Melakukan Prediksi pada Data Testing

Pada tahap ini, model Gaussian Naïve Bayes yang telah dilatih digunakan untuk memprediksi spesies bunga pada data testing. Fungsi `predict()` menerima `X_test` sebagai input dan menghasilkan prediksi label untuk setiap data.

Hasil prediksi disimpan dalam variabel `result`. Nilai yang dihasilkan berupa label numerik yang merepresentasikan spesies bunga Iris. Hasil ini nantinya dapat dibandingkan dengan label sebenarnya pada `y_test` untuk mengetahui ketepatan klasifikasi model.

```python
result = NB_model.predict(X_test)
result
```

### 9. Menghitung Akurasi Data Testing

Pada tahap ini, fungsi `accuracy_score()` dari Scikit-learn digunakan untuk mengevaluasi kinerja model Gaussian Naïve Bayes. Fungsi ini membandingkan label sebenarnya pada data testing (`y_test`) dengan hasil prediksi model (`result`).

Hasil perbandingan tersebut menunjukkan persentase data testing yang berhasil diklasifikasikan dengan benar. Nilai akurasi dikalikan dengan 100 agar hasilnya ditampilkan dalam bentuk persentase. Nilai ini digunakan untuk mengetahui ketepatan model dalam mengklasifikasikan spesies bunga Iris.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

### 10. Prediksi Spesies pada Data Baru

Pada tahap ini, model Gaussian Naïve Bayes yang telah dilatih digunakan untuk memprediksi spesies bunga Iris berdasarkan data baru. Data input yang digunakan adalah `[5.1, 3.5, 1.4, 0.2]`, yang mewakili panjang sepal, lebar sepal, panjang petal, dan lebar petal.

Fungsi `predict()` digunakan untuk menghasilkan prediksi dalam bentuk label numerik. Selanjutnya, fungsi `inverse_transform()` dari `LabelEncoder` mengubah label numerik tersebut menjadi nama spesies bunga agar hasilnya lebih mudah dipahami. Hasil prediksi kemudian ditampilkan menggunakan fungsi `print()`.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

### 11. Prediksi Data Baru Menggunakan DataFrame

Pada tahap ini, data pengukuran bunga baru diubah menjadi DataFrame menggunakan Pandas. Fungsi `np.array()` digunakan untuk membentuk data dalam format array, sedangkan `pd.DataFrame()` mengubahnya menjadi tabel dengan nama kolom yang sesuai dengan fitur pada dataset Iris.

Selanjutnya, model Gaussian Naïve Bayes digunakan untuk memprediksi spesies bunga berdasarkan data tersebut. Hasil prediksi berupa label numerik kemudian dikonversi kembali menjadi nama spesies menggunakan fungsi `inverse_transform()` dari `LabelEncoder`. Nama spesies hasil prediksi ditampilkan sebagai output program.
