### Tugas 1–2: Prediksi Draft Atlet
Memprediksi apakah atlet masuk draft (yes/no)
berdasarkan fitur speed dan agility menggunakan algoritma KNN.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("athletes.xls")

print(df)
```

```python
# Fitur: speed dan agility
feature_columns = ["speed", "agility"]

X = df[feature_columns].values
y = df["draft"].values

print("Fitur (X):")
print(X)

print("\nTarget (y):")
print(y)
```

```python
en = LabelEncoder()
y = en.fit_transform(y)

print("Label hasil encoding:")
print(y)

print("\nUrutan kelas:", en.classes_)
```

```python
X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=0
)

y_train, y_test = train_test_split(
    y,
    test_size=0.2,
    random_state=0
)
```

```python
knn_athletes = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn_athletes.fit(X_train, y_train)

print("Model KNN Athletes berhasil dilatih.")
```

```python
input_athletes = np.array([
    [7.5, 7.0],
    [4.0, 6.5]
])

prediksi = knn_athletes.predict(input_athletes)

hasil = en.inverse_transform(prediksi)

for i in range(len(input_athletes)):
    print(f"Input {i + 1}: {input_athletes[i]}")
    print(f"Prediksi draft: {hasil[i]}")
    print()
```

```python
iris = load_iris()

X_iris = iris.data
y_iris = iris.target

print("Fitur Iris:")
print(iris.feature_names)

print("\nNama spesies:")
print(iris.target_names)
```

```python
X_train_iris, X_test_iris = train_test_split(
    X_iris,
    test_size=0.2,
    random_state=0
)

y_train_iris, y_test_iris = train_test_split(
    y_iris,
    test_size=0.2,
    random_state=0
)
```

```python
knn_iris = KNeighborsClassifier(
    n_neighbors=3,
    metric="euclidean"
)

knn_iris.fit(X_train_iris, y_train_iris)

print("Model KNN Iris berhasil dilatih.")
```

```python
input_iris = np.array([
    [7.1, 1.2, 5.0, 2.0],
    [5.1, 3.5, 1.4, 0.2],
    [6.4, 3.7, 4.2, 2.2]
])

prediksi_iris = knn_iris.predict(input_iris)

for i in range(len(input_iris)):
    spesies = iris.target_names[prediksi_iris[i]]

    print(f"Input {i + 1}: {input_iris[i]}")
    print(f"Prediksi spesies: {spesies}")
    print()
```

### Hasil Prediksi Dataset Iris

Berdasarkan pengujian menggunakan algoritma K-Nearest
Neighbors (KNN) dengan nilai K = 3 dan metrik jarak
Euclidean, diperoleh hasil sebagai berikut:

1. Input [7.1, 1.2, 5.0, 2.0] diprediksi sebagai virginica.
2. Input [5.1, 3.5, 1.4, 0.2] diprediksi sebagai setosa.
3. Input [6.4, 3.7, 4.2, 2.2] diprediksi sebagai versicolor.

Hasil tersebut menunjukkan bahwa model KNN dapat
