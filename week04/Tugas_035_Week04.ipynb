### Tugas 1–2: Prediksi Draft Atlet
Memprediksi apakah atlet masuk draft (yes/no)
berdasarkan fitur speed dan agility menggunakan algoritma KNN.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

Cell ini digunakan untuk mengimpor berbagai pustaka (libraries) dan modul penting yang dibutuhkan dalam proses pengolahan data dan pembuatan model machine learning. Pustaka yang diimpor meliputi numpy untuk komputasi numerik, pandas untuk manipulasi data, matplotlib.pyplot untuk visualisasi data, serta beberapa modul dari scikit-learn seperti pemuatan dataset (load_iris), pembagian data (train_test_split), pengkodean label (LabelEncoder), algoritma klasifikasi (KNeighborsClassifier), dan metrik evaluasi akurasi (accuracy_score).

```python
df = pd.read_csv("athletes.xls")

print(df)
```

Cell ini digunakan untuk membaca dataset dari sebuah file (dalam contoh ini menggunakan fungsi pd.read_csv() untuk membaca file "athletes.xls") ke dalam bentuk DataFrame Pandas dan menyimpannya ke dalam variabel df. Perintah print(df) kemudian digunakan untuk menampilkan keseluruhan atau sebagian isi tabel DataFrame tersebut ke layar.

```python
# Fitur: speed dan agility
feature_columns = ["speed", "agility"]

X = df[feature_columns].values
y = df["draft"].values

print("Fitur (X):")
print(X)

print("\nTarget (y):")
print(y)
```

Cell ini digunakan untuk memisahkan data ke dalam variabel fitur (X) dan variabel target (y) yang akan dipakai dalam proses machine learning. Variabel feature_columns memilih kolom "speed" dan "agility" sebagai data fitur X, sedangkan kolom "draft" diambil sebagai target atau label y. Perintah print kemudian dijalankan untuk menampilkan isi dari data fitur X dan target y ke layar.

```python
en = LabelEncoder()
y = en.fit_transform(y)

print("Label hasil encoding:")
print(y)

print("\nUrutan kelas:", en.classes_)
```

Cell ini digunakan untuk mengubah label target y yang berbentuk teks ('no' dan 'yes') menjadi nilai numerik menggunakan LabelEncoder dari Scikit-learn. Objek en dibuat untuk melakukan proses encoding, kemudian metode fit_transform(y) mempelajari label yang tersedia dan mengubahnya menjadi angka 0 dan 1. Perintah print kemudian menampilkan hasil label yang telah diubah ke dalam bentuk array numerik beserta urutan kelas aslinya yaitu ['no', 'yes'].

```python
X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=0
)

y_train, y_test = train_test_split(
    y,
    test_size=0.2,
    random_state=0
)
```

Cell ini digunakan untuk membagi dataset fitur (X) dan target (y) menjadi data latih (training data) dan data uji (testing data) menggunakan fungsi train_test_split dari Scikit-learn. Parameter test_size=0.2 menetapkan 20% data sebagai data uji dan 80% sisanya sebagai data latih, sementara random_state=0 memastikan pembagian data bersifat konsisten dan dapat direproduksi.

```python
knn_athletes = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn_athletes.fit(X_train, y_train)

print("Model KNN Athletes berhasil dilatih.")
```

Cell ini digunakan untuk menginisialisasi model K-Nearest Neighbors (KNN) melalui kelas KNeighborsClassifier dengan parameter n_neighbors=5 (menggunakan 5 tetangga terdekat) dan metric="euclidean" untuk menghitung jarak antar data. Metode knn_athletes.fit(X_train, y_train) kemudian dijalankan untuk melatih model menggunakan data latih, dan perintah print menampilkan pesan konfirmasi bahwa model berhasil dilatih.

```python
input_athletes = np.array([
    [7.5, 7.0],
    [4.0, 6.5]
])

prediksi = knn_athletes.predict(input_athletes)

hasil = en.inverse_transform(prediksi)

for i in range(len(input_athletes)):
    print(f"Input {i + 1}: {input_athletes[i]}")
    print(f"Prediksi draft: {hasil[i]}")
    print()
```

Cell ini digunakan untuk melakukan prediksi data baru (input_athletes) menggunakan model KNN yang telah dilatih. Data input berupa array NumPy dengan dua sampel data atlet (kecepatan dan kelincahan) diproses oleh metode knn_athletes.predict() untuk menghasilkan prediksi numerik. Nilai numerik tersebut kemudian dikembalikan ke bentuk teks aslinya ('yes' atau 'no') menggunakan metode en.inverse_transform(), dan perulangan for menampilkan hasil prediksi untuk masing-masing data input ke layar.

```python
iris = load_iris()

X_iris = iris.data
y_iris = iris.target

print("Fitur Iris:")
print(iris.feature_names)

print("\nNama spesies:")
print(iris.target_names)
```

Cell ini digunakan untuk memuat dataset bawaan Scikit-learn yaitu dataset Iris menggunakan fungsi load_iris(). Variabel X_iris diisi dengan data fitur pengukuran bunga, sedangkan y_iris diisi dengan label kelas targetnya. Perintah print kemudian dijalankan untuk menampilkan daftar nama fitur (feature_names) serta nama-nama spesies bunga Iris (target_names) ke layar.

```python
X_train_iris, X_test_iris = train_test_split(
    X_iris,
    test_size=0.2,
    random_state=0
)

y_train_iris, y_test_iris = train_test_split(
    y_iris,
    test_size=0.2,
    random_state=0
)
```

Cell ini digunakan untuk membagi dataset Iris (X_iris dan y_iris) menjadi data latih (training data) dan data uji (testing data) menggunakan fungsi train_test_split dari Scikit-learn. Parameter test_size=0.2 menetapkan 20% data sebagai data uji dan 80% sisanya sebagai data latih, sementara random_state=0 memastikan pembagian data bersifat konsisten dan dapat direproduksi.

```python
knn_iris = KNeighborsClassifier(
    n_neighbors=3,
    metric="euclidean"
)

knn_iris.fit(X_train_iris, y_train_iris)

print("Model KNN Iris berhasil dilatih.")
```

Cell ini digunakan untuk menginisialisasi model K-Nearest Neighbors (KNN) khusus untuk dataset Iris melalui kelas KNeighborsClassifier dengan parameter n_neighbors=3 (menggunakan 3 tetangga terdekat) dan metric="euclidean" untuk menghitung jarak. Metode knn_iris.fit(X_train_iris, y_train_iris) kemudian dijalankan untuk melatih model menggunakan data latih tersebut, dan perintah print menampilkan pesan bahwa model KNN untuk Iris berhasil dilatih.

```python
input_iris = np.array([
    [7.1, 1.2, 5.0, 2.0],
    [5.1, 3.5, 1.4, 0.2],
    [6.4, 3.7, 4.2, 2.2]
])

prediksi_iris = knn_iris.predict(input_iris)

for i in range(len(input_iris)):
    spesies = iris.target_names[prediksi_iris[i]]

    print(f"Input {i + 1}: {input_iris[i]}")
    print(f"Prediksi spesies: {spesies}")
    print()
```

Cell ini digunakan untuk melakukan prediksi spesies bunga Iris berdasarkan data pengukuran baru (input_iris) menggunakan model KNN yang telah dilatih sebelumnya. Data input yang berupa array NumPy berisi 3 sampel data pengukuran diproses oleh metode knn_iris.predict() untuk menghasilkan tebakan kelas numerik. Perulangan for kemudian menerjemahkan hasil prediksi angka tersebut ke dalam nama spesies aslinya menggunakan iris.target_names, lalu menampilkan data input beserta prediksi spesiesnya ke layar (seperti virginica, setosa, dan versicolor).
