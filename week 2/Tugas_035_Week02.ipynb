### Analisis Statistik Deskriptif Previous Purchases

Kolom `Previous Purchases` digunakan untuk menganalisis jumlah pembelian
yang telah dilakukan oleh pelanggan sebelumnya.

Analisis statistik deskriptif dilakukan untuk mengetahui karakteristik
data melalui nilai mean, standard deviation (std), minimum, Q1, Q2, Q3,
dan maksimum.

```python
previous_purchases = shopping_df["Previous Purchases"]

print("Mean :", previous_purchases.mean())
print("Std  :", previous_purchases.std())
print("Min  :", previous_purchases.min())
print("Q1   :", previous_purchases.quantile(0.25))
print("Q2   :", previous_purchases.quantile(0.50))
print("Q3   :", previous_purchases.quantile(0.75))
print("Max  :", previous_purchases.max())
```

### Menghitung Statistik Previous Purchases

Kode berikut digunakan untuk menghitung nilai statistik pada kolom
`Previous Purchases` secara terpisah.

Fungsi `mean()` digunakan untuk menghitung rata-rata, `std()` untuk
menghitung standar deviasi, `min()` untuk mendapatkan nilai minimum,
dan `max()` untuk mendapatkan nilai maksimum.

Fungsi `quantile()` digunakan untuk mendapatkan nilai kuartil.
Q1 menggunakan kuantil 0.25, Q2 atau median menggunakan kuantil 0.50,
dan Q3 menggunakan kuantil 0.75.

```python
age_purchase = shopping_df[["Age", "Purchase Amount (USD)"]]

age_purchase.corr()
```

### Analisis Korelasi Age dan Purchase Amount (USD)

Pada bagian ini dipilih kolom `Age` dan `Purchase Amount (USD)` untuk
dianalisis menggunakan korelasi.

Analisis ini bertujuan untuk mengetahui hubungan linear antara usia
pelanggan dengan jumlah pembelian (`Purchase Amount (USD)`).

```python
corr_age_purchase = age_purchase.corr().loc[
    "Age", "Purchase Amount (USD)"
]

print("Korelasi Age vs Purchase Amount (USD):")
print(corr_age_purchase)
```

### Menampilkan Nilai Korelasi Age dan Purchase Amount

Kode berikut digunakan untuk mengambil dan menampilkan nilai korelasi antara
`Age` dan `Purchase Amount (USD)`.

```python
purchase_review = shopping_df[
    ["Purchase Amount (USD)", "Review Rating"]
]

purchase_review.corr()
```

### Menghitung Korelasi Purchase Amount dan Review Rating

Kode berikut digunakan untuk menghitung korelasi antara
`Purchase Amount (USD)` dan `Review Rating` menggunakan fungsi `corr()`.

```python
corr_purchase_review = purchase_review.corr().loc[
    "Purchase Amount (USD)", "Review Rating"
]

print("Korelasi Purchase Amount (USD) vs Review Rating:")
print(corr_purchase_review)
```

### Menampilkan Nilai Korelasi Purchase Amount dan Review Rating
