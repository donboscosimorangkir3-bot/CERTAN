```python
import pandas as pd

path = "data/Titanic-Dataset.csv"
df_titanic = pd.read_csv(path)

df_titanic.head()
```

### Import Library dan Dataset

Mengimpor library pandas dan membaca dataset `Titanic-Dataset.csv` untuk melakukan proses handling missing value.

```python
df_titanic.isnull().sum()
```

### Mengecek Missing Value

Menghitung jumlah nilai yang hilang pada setiap kolom menggunakan `isnull().sum()`.

```python
df_titanic['Age'] = df_titanic['Age'].fillna(
    df_titanic['Age'].median()
)
```

### Menangani Missing Value pada Age

Missing value pada kolom `Age` diisi menggunakan nilai median karena `Age` merupakan data numerik dan median lebih tahan terhadap pengaruh nilai ekstrem.

```python
df_titanic['Embarked'] = df_titanic['Embarked'].fillna(
    df_titanic['Embarked'].mode()[0]
)
```

### Menangani Missing Value pada Embarked

Missing value pada kolom `Embarked` diisi menggunakan nilai yang paling sering muncul atau modus.

```python
df_titanic['Cabin'] = df_titanic['Cabin'].fillna('Unknown')
```

### Menangani Missing Value pada Cabin

Missing value pada kolom `Cabin` diisi dengan nilai `Unknown` untuk menandai bahwa informasi kabin tidak tersedia.

```python
df_titanic.isnull().sum()
```

### Mengecek Missing Value Setelah Ditangani

Menampilkan kembali jumlah missing value untuk memastikan seluruh missing value pada dataset telah ditangani.

### Kesimpulan

- Missing value pada `Age` ditangani menggunakan **median**.
- Missing value pada `Embarked` ditangani menggunakan **modus**.
- Missing value pada `Cabin` diisi dengan **`Unknown`**.
- Pemeriksaan kembali menggunakan `df_titanic.isnull().sum()` dilakukan untuk memastikan missing value telah ditangani.
