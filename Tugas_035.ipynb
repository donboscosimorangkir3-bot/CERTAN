```python
# 1. Matriks 5x5 random, nilai max, min, & rata-rata
import numpy as np

# Membuat matriks random berukuran 5x5
matriks = np.random.randint(1, 101, size=(5, 5))

print("Matriks 5x5:")
print(matriks)

# Mencari nilai maksimum, minimum, dan rata-rata
nilai_maksimum = np.max(matriks)
nilai_minimum = np.min(matriks)
nilai_rata_rata = np.mean(matriks)

print("\nNilai maksimum :", nilai_maksimum)
print("Nilai minimum  :", nilai_minimum)
print("Nilai rata-rata:", nilai_rata_rata)
```

```python
# 2. Grafik 2 Dimensi (Persamaan Implisit)
import numpy as np
import matplotlib.pyplot as plt

# Membuat rentang nilai x dan y
x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)

# Membuat grid 2 dimensi
X, Y = np.meshgrid(x, y)

# Persamaan
Z = (
    X**6
    + 3*X**4*Y**2
    - 3*X**4
    + 3*X**2*Y**4
    - X**2*Y**3
    - 6*X**2*Y**2
    + 3*X**2
    + Y**6
    - 3*Y**4
    + 3*Y**2
)

# Membuat grafik 2 dimensi
plt.figure(figsize=(8, 8))
plt.contour(X, Y, Z, levels=[1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Grafik 2 Dimensi")
plt.grid(True)
plt.axis("equal")

plt.show()
```

```python
# 3. Grafik 3 Dimensi
import numpy as np
import matplotlib.pyplot as plt

# Membuat rentang nilai x dan y
x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)

# Membuat grid 2 dimensi
X, Y = np.meshgrid(x, y)

# Persamaan
Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

# Membuat grafik 3 dimensi
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X, Y, Z)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("Grafik 3 Dimensi")

plt.show()
```
